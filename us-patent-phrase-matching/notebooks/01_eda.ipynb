# U.S. Patent Phrase to Phrase Matching
## 01 - Exploratory Data Analysis

Understand the data before building models.

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Settings
plt.style.use('ggplot')
sns.set_palette('husl')
```

```python
# Define paths
PROJECT_ROOT = Path('../..')
DATA_DIR = PROJECT_ROOT / 'data'
print(f'Data directory: {DATA_DIR}')
```

```python
# Load data
train_df = pd.read_csv(DATA_DIR / 'train.csv')
test_df = pd.read_csv(DATA_DIR / 'test.csv')
sample_sub = pd.read_csv(DATA_DIR / 'sample_submission.csv')

print(f'Train shape: {train_df.shape}')
print(f'Test shape: {test_df.shape}')
print(f'Sample submission shape: {sample_sub.shape}')
```

## Train Data Overview

```python
# Train data info
print('Train columns:', train_df.columns.tolist())
print('\nFirst 5 rows:')
train_df.head()
```

```python
# Test data info
print('Test columns:', test_df.columns.tolist())
print('\nFirst 5 rows:')
test_df.head()
```

```python
# Missing values
print('Missing values in train:')
print(train_df.isnull().sum())
print('\nMissing values in test:')
print(test_df.isnull().sum())
```

## Target Variable Analysis

```python
# Score distribution
print('Score statistics:')
print(train_df['score'].describe())
```

```python
# Plot score distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(train_df['score'], bins=50, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Score')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Score Distribution')

# Box plot
axes[1].boxplot(train_df['score'])
axes[1].set_ylabel('Score')
axes[1].set_title('Score Box Plot')

plt.tight_layout()
plt.show()
```

## Context (CPC Class) Analysis

```python
# CPC class distribution
context_counts = train_df['context'].value_counts()
print(f'Number of unique contexts: {len(context_counts)}')
print('\nTop 15 contexts:')
print(context_counts.head(15))
```

```python
# Plot context distribution (top 20)
fig, ax = plt.subplots(figsize=(12, 6))
context_counts.head(20).plot(kind='bar', ax=ax)
ax.set_xlabel('Context (CPC Class)')
ax.set_ylabel('Count')
ax.set_title('Top 20 CPC Context Classes')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Text Length Analysis

```python
# Text length analysis
train_df['anchor_len'] = train_df['anchor'].apply(len)
train_df['target_len'] = train_df['target'].apply(len)
train_df['context_len'] = train_df['context'].apply(len)

print('Text length statistics:')
print('Anchor:', train_df['anchor_len'].describe())
print('\nTarget:', train_df['target_len'].describe())
print('\nContext:', train_df['context_len'].describe())
```

```python
# Plot text lengths
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].hist(train_df['anchor_len'], bins=30, alpha=0.7)
axes[0].set_title('Anchor Length')

axes[1].hist(train_df['target_len'], bins=30, alpha=0.7)
axes[1].set_title('Target Length')

axes[2].hist(train_df['context_len'], bins=30, alpha=0.7)
axes[2].set_title('Context Length')

plt.tight_layout()
plt.show()
```

## Score by Context Analysis

```python
# Average score by context
score_by_context = train_df.groupby('context')['score'].agg(['mean', 'std', 'count'])
score_by_context = score_by_context.sort_values('count', ascending=False).head(20)
print('Score by context (top 20):')
score_by_context
```

## Key Findings

```python
# Summary
print('=' * 50)
print('KEY FINDINGS')
print('=' * 50)
print(f'Train samples: {len(train_df):,}')
print(f'Test samples: {len(test_df):,}')
print(f'Unique contexts: {train_df["context"].nunique()}')
print(f'Score range: {train_df["score"].min():.2f} - {train_df["score"].max():.2f}')
print(f'Score mean: {train_df["score"].mean():.3f}')
print(f'Score std: {train_df["score"].std():.3f}')
```

```python
# Save analysis summary
summary = {
    'train_samples': len(train_df),
    'test_samples': len(test_df),
    'unique_contexts': train_df['context'].nunique(),
    'score_mean': train_df['score'].mean(),
    'score_std': train_df['score'].std(),
    'score_min': train_df['score'].min(),
    'score_max': train_df['score'].max()
}
import json
with open(PROJECT_ROOT / 'analysis_summary.json', 'w') as f:
    json.dump(summary, f, indent=2)
print('Summary saved to analysis_summary.json')
```
